# 01. AI-AgriVision: Dataset Analysis & Exploration
This notebook performs exploratory data analysis (EDA) on the crop leaf disease dataset.
It inspects class distributions, image dimensions, color profiles, and detects sample imbalances.

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from collections import Counter

DATASET_PATH = '../dataset/train'

classes = [d for d in os.listdir(DATASET_PATH) if os.path.isdir(os.path.join(DATASET_PATH, d))]
print(f"Total classes: {len(classes)}")

counts = {}
for c in classes:
    files = glob.glob(os.path.join(DATASET_PATH, c, '*.*'))
    counts[c] = len(files)

plt.figure(figsize=(12, 6))
plt.bar(counts.keys(), counts.values(), color='forestgreen')
plt.xticks(rotation=90)
plt.title('Crop Disease Class Distribution')
plt.ylabel('Number of Samples')
plt.show()
```
