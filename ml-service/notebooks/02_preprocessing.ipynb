# 02. AI-AgriVision: Preprocessing & Data Augmentation Pipeline
Inspects 224x224 RGB image normalization, Laplacian blur calculation, and geometric/photometric augmentations.

```python
import cv2
import numpy as np
from PIL import Image
import tensorflow as tf

def inspect_blur(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    laplacian_var = cv2.Laplacian(img, cv2.CV_64F).var()
    return laplacian_var

print("Blur threshold set at 80.0. Scores below 80 are flagged for re-capture.")
```
